```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
resize = 1
plt.rcParams.update({
    "figure.figsize": (6.4*resize, 4.0*resize), # (6.4, 4.8)[4:3] -> (6.4, 4.0)[8:5]
    "font.sans-serif": ["Helvetica", "Nimbus Sans", "Arial", "DejaVu Sans"],
})
```

```python
dir_path = "./npy"

def load_npy(filename: str):
    return np.load(filename, allow_pickle=True).item()
```

```python
# load file and print items
#ff = load_npy("./npy/imagenet_resnet18_e105_t105_w4s1_amp_schedule-cyclic.npy")
#for k, v in ff.items():
#    print(k, v[-5:])
```

```python
w2s0 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w2s0_amp.npy")
#w3s0 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w3s0_amp.npy")
w3s1 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w3s1_amp.npy")
w3s2 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w3s2_amp.npy")
#w4s0 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w4s0_amp.npy")
#w4s1 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w4s1_amp.npy")
w4s2 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w4s2_amp.npy")
w4s3 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w4s3_amp.npy")

w5s0 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w5s0_amp.npy")
w5s1 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w5s1_amp.npy")
w5s2 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w5s2_amp.npy")
w5s3 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w5s3_amp.npy")
w5s4 = load_npy(f"{dir_path}/imagenet_resnet18_e105_t105_w5s4_amp.npy")
```

```python
label = {
    "1 worker, 0S1L": w2s0,
    #"2 workers, 0S2L": w3s0,
    "2 workers, 1S1L": w3s1,
    #"2 workers, 2S0L": w3s2,
    #"3 workers, 0S3L": w4s0,
    #"3 workers, 1S2L": w4s1,
    "3 workers, 2S1L": w4s2,
    #"3 workers, 3S0L": w4s3,
    "4 workers, 0S4L": w5s0,
    "4 workers, 1S3L": w5s1,
    "4 workers, 2S2L": w5s2,
    "4 workers, 3S1L": w5s3,
    "4 workers, 4S0L": w5s4,
}
```

```python
print(w3s1.keys())
```

```python
plt.figure()

for k, v in label.items():
    plt.plot(v["commit_time"], v["val_acc"], label=f"{k} ({v['val_acc'][-1]:.3f})")

plt.xlabel('Time (sec)')
plt.ylabel('Accuracy')
plt.legend()
```

```python
# filter by num of cluster
plt.figure()

for k, v in label.items():
    if "3 workers" in k:
        plt.plot(v["commit_time"], v["val_acc"], label=f"{k} ({v['val_acc'][-1]:.3f})")

plt.xlabel('Time (sec)')
plt.ylabel('Accuracy')
plt.legend()
```

```python
# small:large = (n-1):1
plt.figure()

plt.plot(w2s0["commit_time"], w2s0["val_acc"], label=f"1 workers, 0S1L (acc: {w2s0['val_acc'][-1]*100:.1f}%, time: {w2s0['commit_time'][-1]:.0f} sec)")
plt.plot(w3s1["commit_time"], w3s1["val_acc"], label=f"2 workers, 1S1L (acc: {w3s1['val_acc'][-1]*100:.1f}%, time: {w3s1['commit_time'][-1]:.0f} sec)")
plt.plot(w4s2["commit_time"], w4s2["val_acc"], label=f"3 workers, 2S1L (acc: {w4s2['val_acc'][-1]*100:.1f}%, time: {w4s2['commit_time'][-1]:.0f} sec)")
plt.plot(w5s3["commit_time"], w5s3["val_acc"], label=f"4 workers, 3S1L (acc: {w5s3['val_acc'][-1]*100:.1f}%, time: {w5s3['commit_time'][-1]:.0f} sec)")

plt.xlabel('Time (sec)')
plt.ylabel('Accuracy')
plt.legend()
```
